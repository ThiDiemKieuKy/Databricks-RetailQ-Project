```sql
CREATE OR REPLACE VIEW retailq_dev.semantic.retailq_metrics
WITH METRICS
LANGUAGE YAML
AS $$
version: 1.1

source: retailq_dev.gold.fact_sales
comment: Retail metrics for analyzing sales performance, customer behavior, and product trends

joins:

  - name: customer
    source: retailq_dev.gold.dim_customer
    on: source.customer_id = customer.customer_id

  - name: product
    source: retailq_dev.gold.dim_product
    on: source.product_id = product.product_id

  - name: calendar
    source: retailq_dev.gold.dim_calendar
    on: source.transaction_date = calendar.date_key

dimensions:
  - name: transaction_date
    expr: calendar.date_key
    display_name: Transaction Date
    comment: Date of the transaction
    synonyms:
      - date
      - transaction date

  - name: year
    expr: calendar.year
    display_name: Year
    comment: Year of the transaction
    synonyms:
      - year
      - transaction year

  - name: month
    expr: calendar.month
    display_name: Month
    comment: Month of the transaction
    synonyms:
      - month
      - transaction month
      - period

  - name: product_category
    expr: product.category
    display_name: Product Category
    comment: Category of the product sold
    synonyms:
      - category
      - product type

  - name: brand
    expr: product.brand
    display_name: Brand
    comment: Brand of the product sold
    synonyms:
      - brand
      - product brand

  - name: customer_type
    expr: customer.customer_type
    display_name: Customer Type
    comment: Type or segment of customer
    synonyms:
      - customer segment
      - account type

  - name: payment_mode
    expr: payment_mode
    display_name: Payment Mode
    comment: Method of payment used for transaction
    synonyms:
      - payment method
      - payment type

  - name: sales_channel
    expr: sales_channel
    display_name: Sales Channel
    comment: Channel through which sale was made
    synonyms:
      - channel
      - distribution channel

  - name: year_month
    expr: calendar.year_month
    display_name: Year-Month
    comment: Year and month of transaction
    synonyms:
      - month
      - transaction month
      - period

  - name: opportunity_stage
    expr: stage_name
    display_name: Opportunity Stage
    comment: Stage of the sales opportunity
    synonyms:
      - stage
      - sales stage

  - name: Customer Name
    expr: customer.customer_name
    display_name: Customer Name
    comment: Name of the customer
    synonyms:
      - customer
      - customer name

  - name: Billing City
    expr: customer.customer_city
    display_name: City
    comment: Customer billing city
    synonyms:
      - city
      - customer city

  - name: Billing State
    expr: customer.customer_state
    display_name: State
    comment: Customer billing state
    synonyms:
      - state
      - customer state

  - name: Billing Country
    expr: customer.customer_country
    display_name: Country
    comment: Customer billing country
    synonyms:
      - country
      - customer country

  - name: Industry
    expr: customer.industry
    display_name: Industry
    comment: Customer industry sector
    synonyms:
      - customer industry
      - sector

measures:

  - name: total_revenue
    expr: SUM(selling_price * quantity)
    display_name: Total Revenue
    comment: Total revenue from all sales transactions
    format:
      type: currency
      currency_code: USD
      decimal_places:
        type: exact
        places: 2
    synonyms:
      - revenue
      - sales
      - total sales

  - name: total_discount
    expr: SUM(discount_amount)
    display_name: Total Discount
    comment: Total discount amount applied to transactions
    format:
      type: currency
      currency_code: USD
      decimal_places:
        type: exact
        places: 2
    synonyms:
      - discount
      - discounts given

  - name: transaction_count
    expr: COUNT(DISTINCT transaction_id)
    display_name: Transaction Count
    comment: Number of unique transactions
    format:
      type: number
      decimal_places:
        type: exact
        places: 0
    synonyms:
      - transactions
      - number of sales
      - order count

  - name: avg_transaction_value
    expr: SUM(selling_price * quantity) / COUNT(DISTINCT transaction_id)
    display_name: Average Transaction Value
    comment: Average revenue per transaction
    format:
      type: currency
      currency_code: USD
      decimal_places:
        type: exact
        places: 2
    synonyms:
      - average order value
      - AOV
      - avg sale

  - name: total_quantity
    expr: SUM(quantity)
    display_name: Total Quantity Sold
    comment: Total number of units sold across all transactions
    format:
      type: number
      decimal_places:
        type: exact
        places: 0
    synonyms:
      - units sold
      - quantity
      - volume

  - name: unique_customers
    expr: COUNT(DISTINCT customer_id)
    display_name: Unique Customers
    comment: Total number of unique customers who made a purchase
    format:
      type: number
      decimal_places:
        type: exact
        places: 0
    synonyms:
      - customers
      - unique customers
      - number of customers
$$
```
